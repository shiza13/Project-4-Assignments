# **Project 2: Guess the Number Game Python Project (computer)**

```python
import random

print("First think of a number between 0 and 100, and I will try to guess it!")
input("Press Enter when you're ready ")

low = 0
high = 100
feedback = ""

while feedback != "c":
    guess = random.randint(low, high)
    print(f"My guess is {guess}")
    feedback = input("Is it (H)igh, (L)ow, or (C)orrect? ").lower()

    if feedback == "h":
        high = guess - 1
    elif feedback == "l":
        low = guess + 1
    elif feedback != "c":
        print("Please enter H, L, or C.")

print(f"Hurray! I won the game, {guess}!")
```

# **Project 3: Guess the Number Game Python Project (user)**

```python
import random

number = random.randint(0, 100)
guess = None
tries = 0

print("I have picked a number between 0 and 100. Try to guess it!")

while guess != number:
    guess = int(input("Your guess: "))
    tries += 1

    if guess < number:
        print("Too low! Try again.")
    elif guess > number:
        print("Too high! Try again.")

print(f"Correct! You guessed it in {tries} tries!")
```

# **Project 4: Rock, paper, scissors Python Project**

```python
import random

options = ["rock", "paper", "scissors"]

print("Hello my friend, let's play Rock, Paper, Scissors!")
user = input("Choose rock, paper, or scissors: ").lower()
computer = random.choice(options)

print(f"\nYou chose: {user}")
print(f"Computer chose: {computer}")

if user == computer:
    print("It's a tie!")
elif (user == "rock" and computer == "scissors") or \
     (user == "paper" and computer == "rock") or \
     (user == "scissors" and computer == "paper"):
    print("You win!")
elif user in options:
    print("Computer wins!")
else:
    print("Invalid input! Please choose rock, paper, or scissors.")
```

# **Project 5: Hangman Python Project**

```python
import random

word_dict = {
    "Fruits": ["apple", "banana", "grape", "mango", "orange"],
    "Animals": ["tiger", "elephant", "monkey", "rabbit", "giraffe"],
    "Vehicles": ["truck", "motorcycle", "airplane", "car", "scooter"]
}


category = random.choice(list(word_dict.keys()))
secret_word = random.choice(word_dict[category])

guessed_letters = []
tries = 6

print("🎮 Welcome to Hangman!")
print(f"Hint: The word is from the category — {category}")
print("_ " * len(secret_word))

while tries > 0:
    guess = input("\nGuess a letter: ").lower()

    if not guess.isalpha() or len(guess) != 1:
        print("Please enter a single alphabet letter.")
        continue

    if guess in guessed_letters:
        print("You already guessed that letter.")
        continue

    guessed_letters.append(guess)

    if guess in secret_word:
        print("✅ Good guess!")
    else:
        print("❌ Wrong guess!")
        tries -= 1

    display_word = ""
    for letter in secret_word:
        if letter in guessed_letters:
            display_word += letter + " "
        else:
            display_word += "_ "
    print(display_word)
    print(f"Tries left: {tries}")

    if all(letter in guessed_letters for letter in secret_word):
        print(f"\n🎉 You guessed it! The word was '{secret_word}'")
        break
else:
    print(f"\n😢 Out of tries! The word was: {secret_word}")
```

# **Project 6: Countdown Timer Python Project**

```python
import time
import os
import platform

minutes = int(input("Enter minutes: "))
seconds = int(input("Enter seconds: "))
total_seconds = minutes * 60 + seconds

for remaining in range(total_seconds, 0, -1):
    mins = remaining // 60
    secs = remaining % 60
    print(f"Time Left: {mins:02d}:{secs:02d}", end='\r')
    time.sleep(1)

print(" " * 20, end="\r")
print("⏰ Timer Done!")

if platform.system() == "Windows":
    import winsound
    winsound.Beep(1000, 1000)
else:
    os.system('say "Time is up"')
```

# **Project 7: Password Generator Python Project**

```python
import random
import string

num_passwords = int(input("How many passwords do you want to generate? "))
password_length = int(input("Enter the length of each password: "))

all_chars = string.ascii_letters + string.digits + string.punctuation

for _ in range(num_passwords):
    password = ''.join(random.choice(all_chars) for _ in range(password_length))
    print(password)
```
